```python
# Task 8: Functions and Lambda
# Create a function to calculate the square of a number.
# Then, use a lambda function to do the same.

# Your code here:
def square(num):
    return num ** 2

lambda_spare = lambda num: num ** 2

inputs = [2, 6]

for n in inputs:
    print(f'(Function) Square of {n} =', square(n))

print('\n')

for n in inputs:
    print(f'(Lamda) Square of {n} =', lambda_spare(n))
```

```python
# Task 9: Iterators and Generators
# Create an iterator for the first 5 even numbers.
# Then, create a generator for the same.

# Your code here:
class EvenIterator:
    def __init__(self):
        self.current = 0   # Start at 0
        self.count = 0     # Track how many even numbers have been produced

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= 5:   # Stop after 5 even numbers
            raise StopIteration
        even_number = self.current
        self.current += 2
        self.count += 1
        return even_number

# Demonstrate iterator
print("Iterator output:")
for num in EvenIterator():
    print(num)
```

```python
# Task 10: Map, Reduce, and Filter
# Use map to square all numbers in a list.
# Use reduce to find the product of all numbers in a list.
# Use filter to get only even numbers from a list.

numbers = [1, 2, 3, 4, 5]

# Your code here:
map_square = list(map(lambda x: x**2, numbers))

reduce_product = reduce(lambda x, y: x * y, numbers)

filter_evens = list(filter(lambda x: x % 2 == 0, numbers))

print('Map: ', map_square)
print('Reduce: ',reduce_product)
print('Filter: ', filter_evens)
```

```python
# Task 11: Object-Oriented Programming - Creating a Class
# Create a class called 'Rectangle' with attributes 'length' and 'width'.
# Include methods to calculate area and perimeter.

# Your code here:
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


# Create Rectangle objects
rect1 = Rectangle(5, 3)
rect2 = Rectangle(10, 4)

# Demonstrate methods
print("Rectangle 1 Length:", rect1.length)
print("Rectangle 1 Width:", rect1.width)
print("Rectangle 1 Area:", rect1.area())
print("Rectangle 1 Perimeter:", rect1.perimeter())

print()
print("Rectangle 2 Length:", rect2.length)
print("Rectangle 2 Width:", rect2.width)
print("Rectangle 2 Area:", rect2.area())
print("Rectangle 2 Perimeter:", rect2.perimeter())
```
