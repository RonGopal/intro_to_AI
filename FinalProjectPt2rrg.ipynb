```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import datetime as dt
from sklearn.preprocessing import OneHotEncoder
import csv
```

```python
hsp = pd.read_csv('historical_stock_prices.csv')
hs = pd.read_csv('historical_stocks.csv')
```

```python
display(hsp.shape)
print()
display(hs.shape)
```

```python
#remove obvious header row
hs = hs[hs['ticker'] !='Symbol'].copy()
```

```python
# convert date column to datetime type
hsp['date'] = pd.to_datetime(hsp['date'])
```

```python
#sort price data
hsp = hsp.sort_values(['ticker', 'date'])
```

```python
df = hsp

df['price_change'] = df['close'].diff()
df['daily_return'] = df.groupby('ticker')['close'].pct_change()
df['close_lag_1'] = df.groupby('ticker')['close'].shift(1)

df['ma5'] = (
    df.groupby('ticker')['close']
    .transform(lambda x: x.rolling(5).mean())
)

df['ma20'] = (
    df.groupby('ticker')['close']
    .transform(lambda x: x.rolling(20).mean())
)
```

```python
df.head()
```

```python
df['volatility_20'] = (
    df.groupby('ticker')['daily_return']
    .transform(lambda x: x.rolling(20).std())
)
```

```python
df.head(30)
```

```python
# Daily price range, how much stock moved in a day

df['daily_range_pct'] = ((df['high'] - df['low']) / df['close'] * 100)
```

```python
df[['ticker', 'date', 'close',
    'price_change', 'daily_return',
    'close_lag_1', 'ma5', 'ma20',
    'volatility_20', 'daily_range_pct']].head(25)
```

```python
metadata = hs[[
    'ticker', 'exchange', 'sector', 'industry'
]]
```

```python
df = df.merge(
    metadata, how='left', on='ticker'
)
```

```python
df['sector'].value_counts()
```

```python
df.head(20)
```

```python
df[['daily_return', 'close_lag_1', 'ma5', 'ma20', 'volatility_20']].isna().sum()
```

```python
df = df.dropna(
    subset=['daily_return', 'close_lag_1', 'ma5', 'ma20', 'volatility_20']
)
```

```python
df.isna().sum()
```

```python
# One-hot encode 'exchange'
exchange_encoder = OneHotEncoder(sparse_output=False)
encoded_exchange = exchange_encoder.fit_transform(df[['exchange']])
exchange_df = pd.DataFrame(encoded_exchange, columns=exchange_encoder.get_feature_names_out(['exchange']), index=df.index)
df = pd.concat([df, exchange_df], axis=1).drop(columns=['exchange'])

# One-hot encode 'sector'
sector_encoder = OneHotEncoder(sparse_output=False)
encoded_sector = sector_encoder.fit_transform(df[['sector']])
sector_df = pd.DataFrame(encoded_sector, columns=sector_encoder.get_feature_names_out(['sector']), index=df.index)
df = pd.concat([df, sector_df], axis=1).drop(columns=['sector'])
```

```python
df.to_csv('stock_prepared.csv', index=False)
```
